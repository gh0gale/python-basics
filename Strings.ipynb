```python
vowels = "aeiou"
n = input("Enter the string")
print(n)
cvc=0
for i in range(1,len(n)-1):
    if n[i]==" ":
        continue
    if n[i-1] not in vowels and n[i] in vowels and n[i+1] not in vowels:
        cvc=cvc + 1
print(cvc)
```
